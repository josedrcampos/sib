# Exercise 6: stratified_train_test_split on iris

```python
import sys
from pathlib import Path

import numpy as np

# find the repository root (the folder that contains "datasets"), wherever the notebook is
ROOT = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / 'datasets').is_dir())
for path in (ROOT / 'src', ROOT):
    if str(path) not in sys.path:
        sys.path.insert(0, str(path))

DATA = ROOT / 'datasets'

from si.io.csv_file import read_csv
from si.model_selection.split import train_test_split, stratified_train_test_split
```

```python
iris = read_csv(str(DATA / 'iris' / 'iris.csv'), features=True, label=True)

def counts(ds):
    labels, n = np.unique(ds.y, return_counts=True)
    return dict(zip(labels, n.tolist()))

print('full:', counts(iris))
```

## Random split vs stratified split
With a random split the class proportions in the test set depend on luck; with the stratified split each class contributes exactly `test_size` of its samples.

```python
train, test = train_test_split(iris, test_size=0.2, random_state=1)
print('random     -> train', counts(train), '| test', counts(test))

train, test = stratified_train_test_split(iris, test_size=0.2, random_state=1)
print('stratified -> train', counts(train), '| test', counts(test))
```

## Using it with a model

```python
from si.models.knn_classifier import KNNClassifier
knn = KNNClassifier(k=3).fit(train)
print('accuracy:', knn.score(test))
```
