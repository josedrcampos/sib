### f_classification
One-way ANOVA for each feature: compares the variance between classes with the variance within classes.
A high F (low p-value) means the feature separates the classes well.

```python
import sys
from pathlib import Path

import pandas as pd

# find the repository root (the folder that contains "datasets"), wherever the notebook is
ROOT = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / 'datasets').is_dir())

# make the "si" package importable without installing it
if str(ROOT / 'src') not in sys.path:
    sys.path.insert(0, str(ROOT / 'src'))

from si.io.csv_file import read_csv
from si.statistics.f_classification import f_classification

IRIS_PATH = ROOT / 'datasets' / 'iris' / 'iris.csv'

iris = read_csv(str(IRIS_PATH), sep=',', features=True, label=True)
iris.shape()
```

```python
F, p = f_classification(iris)

pd.DataFrame({'F': F, 'p-value': p}, index=iris.features).sort_values('F', ascending=False)
```
